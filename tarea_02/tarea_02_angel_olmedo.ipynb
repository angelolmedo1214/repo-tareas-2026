## Problema 1: Clasificador de temperatura

```python
# Recorremos todas las temperaturas desde -30 hasta 50, incluidos ambos extremos.
for temperatura in range(-30, 51):
    # Clasificamos cada temperatura según su valor.
    if temperatura < 0:
        print(temperatura, "Muy frío")
    elif temperatura < 25:
        print(temperatura, "Templado")
    else:
        print(temperatura, "Caluroso")
```

## Problema 2: Números especiales del robot

```python
# Recorremos los números del 1 al 100, incluidos ambos extremos.
for numero in range(1, 101):
    # Primero comprobamos si es divisible por 10.
    if numero % 10 == 0:
        print(numero, "Bip Bip")
    # Los demás múltiplos de 5 reciben un solo «Bip».
    elif numero % 5 == 0:
        print(numero, "Bip")
```

## Problema 3: Detector de puntaje gamer
Ejecuta la celda e ingresa los dos puntajes cuando se soliciten.

```python
# Pedimos dos puntajes y convertimos las respuestas a números enteros.
primer_puntaje = int(input("Ingresa primer puntaje: "))
segundo_puntaje = int(input("Ingresa segundo puntaje: "))

# Sumamos ambos puntajes.
suma = primer_puntaje + segundo_puntaje

# Mostramos el nivel que corresponde al resultado.
if suma < 50:
    print("Resultado: Nivel principiante")
elif suma < 80:
    print("Resultado: Nivel intermedio")
else:
    print("Resultado: Nivel pro")
```

## Problema 4: Calculador de perfil musical
Escribe `si` o `no` en las preguntas sobre música y baile.

```python
# Recibimos la edad y los gustos de la persona.
edad = int(input("Ingresa tu edad: "))
le_gusta_musica = input("¿Te gusta la música? (si/no): ").strip().lower() == "si"
le_gusta_bailar = input("¿Te gusta bailar? (si/no): ").strip().lower() == "si"

# Revisamos primero si cumple las tres condiciones para ser estrella.
if edad >= 18 and le_gusta_musica and le_gusta_bailar:
    print("Eres una estrella de la fiesta")
# Esta opción corresponde a quien disfruta la música, pero no bailar.
elif le_gusta_musica and not le_gusta_bailar:
    print("Te gusta la música, pero no tanto bailar")
# Todos los demás casos tienen un estilo propio.
else:
    print("Tienes tu propio estilo")
```

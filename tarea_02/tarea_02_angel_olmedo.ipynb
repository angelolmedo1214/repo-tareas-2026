```python
# ==========================================
# Problema 1. Clasificador de temperatura
# ==========================================
# Usamos un ciclo for para recorrer los números desde -30 hasta 50.
# En Python, el límite superior del range() no se incluye, por eso usamos 51.
for temp in range(-30, 51):
    if temp < 0:
        print(f"{temp} Muy frío")
    elif 0 <= temp <= 24:
        print(f"{temp} Templado")
    elif temp >= 25:
        print(f"{temp} Caluroso")

print("\n" + "-"*40 + "\n") # Separador visual

# ==========================================
# Problema 2. Números especiales del robot
# ==========================================
# Iteramos del 1 al 100.
for num in range(1, 101):
    # La clave aquí es el orden: primero debemos preguntar si es divisible
    # por 10. Si preguntáramos por el 5 primero, el 10 también caería ahí.
    if num % 10 == 0:
        print(f"{num} Bip Bip")
    elif num % 5 == 0:
        print(f"{num} Bip")

print("\n" + "-"*40 + "\n") # Separador visual

# ==========================================
# Problema 3. Detector de puntaje gamer
# ==========================================
# Solicitamos los datos al usuario usando input().
# Usamos int() para convertir el texto ingresado a un número entero.
puntaje_1 = int(input("Ingresa primer puntaje: "))
puntaje_2 = int(input("Ingresa segundo puntaje: "))

suma = puntaje_1 + puntaje_2

# Evaluamos el resultado de la suma según las reglas dadas.
if suma < 50:
    print(f"\nResultado: Nivel principiante")
elif 50 <= suma < 80:
    print(f"\nResultado: Nivel intermedio")
elif suma >= 80:
    print(f"\nResultado: Nivel pro")

print("\n" + "-"*40 + "\n") # Separador visual

# ==========================================
# Problema 4. Calculador de perfil musical
# ==========================================
# Como las instrucciones piden "recibir parámetros" y "retornar",
# la mejor práctica es crear una función.
def calcular_perfil(edad, le_gusta_musica, le_gusta_bailar):
    # Evaluamos las 3 condiciones simultáneas
    if edad >= 18 and le_gusta_musica and le_gusta_bailar:
        return "Eres una estrella de la fiesta"
    # Evaluamos la segunda regla
    elif le_gusta_musica and not le_gusta_bailar:
        return "Te gusta la música, pero no tanto bailar"
    # Cualquier otro caso (else)
    else:
        return "Tienes tu propio estilo"

# Ejemplos de uso de los parámetros para probar la función:
resultado = calcular_perfil(edad=20, le_gusta_musica=True, le_gusta_bailar=False)
print(resultado)
```
